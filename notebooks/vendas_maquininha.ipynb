```python

#Conversão de tipo dos arquivos
def converter_tipos_maq(df, formato_data, decimal):
    df = df.copy()
    bruto = df.copy()   # texto original, para conferir falhas

    # Datas
    for col in ["Data", "Data recebimento"]:
        df[col] = pd.to_datetime(df[col].astype(str).str.strip(),
                                 format=formato_data, errors="coerce")

    # Valores
    for col in ["Valor bruto", "Taxa", "Valor liquido"]:
        texto = (df[col].astype(str).str.strip()
                        .str.replace("R$", "", regex=False)
                        .str.replace(" ", "", regex=False))
        if decimal == ",":   # formato BR: 1.234,56
            texto = (texto.str.replace(".", "", regex=False)
                          .str.replace(",", ".", regex=False))
        df[col] = pd.to_numeric(texto, errors="coerce")

    # Falha = tinha texto no original, mas virou vazio na conversão
    cols = ["Data", "Data recebimento", "Valor bruto", "Taxa", "Valor liquido"]
    tinha_texto = bruto[cols].apply(lambda s: s.notna() & (s.astype(str).str.strip() != ""))
    falhas = (tinha_texto & df[cols].isna()).sum()
    print("Falhas de conversão por coluna:")
    print(falhas)
    return df

dt25 = converter_tipos_maq(dt25, "%d/%m/%Y", decimal=",")
dt26 = converter_tipos_maq(dt26, "%Y-%m-%d", decimal=".")
```

```python
#Unificação dos dados
dados_maq = {
    "vendas_2025.csv": dt25,
    "relatorio_vendas_jan-out_2026.csv": dt26,
}
maquininha = (pd.concat(dados_maq, names=["arquivo_origem"])
                .reset_index(level=0)
                .reset_index(drop=True))
```

```python
#verificação de unificação
print(maquininha.shape, "=", len(dt25), "+", len(dt26))
print(maquininha.dtypes)

# se a média de um ano for ~100x a do outro, algum decimal foi convertido errado
print(maquininha.groupby("arquivo_origem")["Valor bruto"].describe())
```

```python
#conferencia de sobreposição de periodo
#como esses aarquivos possuem ID, basta verificar se ha algum id repetido
# período de cada arquivo
print(maquininha.groupby("arquivo_origem")["Data"].agg(["min", "max", "count"]))

# verifica se um id aparece mais de uma vez
dup = maquininha[maquininha["Id"].duplicated(keep=False)]
print("Ids repetidos:", dup["Id"].nunique())
dup.sort_values("Id").head(10)
```

```python
#verificação de lacunas de vendas nos periodos. 

# todos os dias do período × dias com venda
todos = pd.date_range(maquininha["Data"].min(), maquininha["Data"].max(), freq="D")# faz uma lista com todos os dias do primeiro ao ultimo
com_venda = maquininha["Data"].dt.normalize().unique()
sem_venda = todos.difference(com_venda)#verifica quias dias não estão na lista de "com vendas"

print("Dias sem venda:", len(sem_venda))

# em que dia da semana caem?
pd.Series(sem_venda.day_name()).value_counts()# transforma as data no nome do dia da semanaa e conta quantas vezes cada dia aparece
```
