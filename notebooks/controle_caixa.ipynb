```python
#verificação de formatos
import re

def ver_formatos(serie, nome):
    """Mostra os formatos de escrita de uma coluna (dígitos viram 9)."""
    s = serie.dropna().astype(str).str.strip()
    formato = s.apply(lambda x: re.sub(r"\d", "9", x))
    resumo = (pd.DataFrame({"formato": formato, "exemplo": s})
                .groupby("formato")
                .agg(qtd=("exemplo", "size"), exemplo=("exemplo", "first"))
                .sort_values("qtd", ascending=False))
    print(f"--- {nome} ---")
    display(resumo)

# 2025: data, entrada e saída
ver_formatos(c25["Data"], "2025 - Data")
ver_formatos(c25["Entrada"], "2025 - Entrada")
ver_formatos(c25["Saída"], "2025 - Saída")

# 2026: data e valor
ver_formatos(c26["Data"], "2026 - Data")
ver_formatos(c26["Valor"], "2026 - Valor")
```

```python
#tratamento de daatas e valores 

# ---------------------------------------------------------------
# 2. LIMPEZA DE LINHAS (2025)
# ---------------------------------------------------------------
c25 = c25.drop(columns=[c for c in c25.columns if str(c).startswith("Unnamed")])  # nota solta (já lida)
eh_total = c25["Descrição"].astype(str).str.strip().str.upper().str.startswith("TOTAL")
print("Linhas de TOTAL removidas:", eh_total.sum())
c25 = c25[~eh_total]
c25 = c25.dropna(how="all")          # linhas totalmente em branco

# ---------------------------------------------------------------
# 3. DATAS
# ---------------------------------------------------------------
def converter_data_caixa(serie, ano):
    s = serie.astype(str).str.strip()
    data = pd.to_datetime(s, format="%Y-%m-%d %H:%M:%S", errors="coerce")       # data do Excel
    sem_ano = s.str.fullmatch(r"\d{1,2}/\d{1,2}")
    data[sem_ano] = pd.to_datetime(s[sem_ano] + f"/{ano}", format="%d/%m/%Y", errors="coerce")
    com_hifen = s.str.fullmatch(r"\d{1,2}-\d{1,2}-\d{4}")
    data[com_hifen] = pd.to_datetime(s[com_hifen], format="%d-%m-%Y", errors="coerce")
    falhas = s[data.isna() & serie.notna()]
    print(f"Datas {ano} - falhas: {len(falhas)}", falhas.unique()[:10])
    return data

c25["Data"] = converter_data_caixa(c25["Data"], 2025)
c26["Data"] = converter_data_caixa(c26["Data"], 2026)

# ---------------------------------------------------------------
# 4. VALORES
# ---------------------------------------------------------------
def converter_valor_caixa(serie):
    texto = (serie.astype(str).str.strip()
                  .str.replace("R$", "", regex=False)
                  .str.replace(" ", "", regex=False)
                  .str.replace(",", ".", regex=False))
    valor = pd.to_numeric(texto, errors="coerce")
    falhas = serie[valor.isna() & serie.notna()]
    print("Valores - falhas:", len(falhas), falhas.unique()[:10])
    return valor

c25["Entrada"] = converter_valor_caixa(c25["Entrada"])
c25["Saída"] = converter_valor_caixa(c25["Saída"])
c26["Valor"] = converter_valor_caixa(c26["Valor"])

# ---------------------------------------------------------------
# 5. 2025 NO PADRÃO DE 2026 (Tipo + Valor)
# ---------------------------------------------------------------
print("Entrada e saída na mesma linha:", (c25["Entrada"].notna() & c25["Saída"].notna()).sum())
print("Sem entrada nem saída:", (c25["Entrada"].isna() & c25["Saída"].isna()).sum())

c25["Tipo"] = np.where(c25["Entrada"].notna(), "Entrada",
              np.where(c25["Saída"].notna(), "Saída", None))
c25["Valor"] = c25["Entrada"].fillna(c25["Saída"])
c25 = c25.drop(columns=["Entrada", "Saída"])

# ---------------------------------------------------------------
# 6. UNIFICAÇÃO + VALOR COM SINAL
# ---------------------------------------------------------------
caixa = (pd.concat({"Caixa 2025": c25, "2026": c26}, names=["aba_origem"])
           .reset_index(level=0)
           .reset_index(drop=True))
caixa["ordem_original"] = range(len(caixa))

caixa["Valor"] = np.where(caixa["Tipo"] == "Saída", -caixa["Valor"].abs(), caixa["Valor"].abs())

print(caixa.shape)
print(caixa.dtypes)
print(caixa["Tipo"].value_counts(dropna=False))
display(caixa.head())
```

```python
#verificando as duas linhas sem entrada e saída
display(caixa[caixa["Tipo"].isna()])
```

```python
#Se foi vendido nesse dia, elas deveriam ter valor. Como não tenho certeza se foi realmente vendido, 
# então vou manter o valor vazio pois 0 atribui valor.
# Atribuir tipo "Entrada" pois esta escrito na descrição que foi venda
#lembrar de verificar essa etapa na conciliação, pois pode ter sido apenas anotado e nao ter tido venda em dinheiero nesse dia

caixa.loc[caixa["Tipo"].isna() & caixa["Descrição"].str.contains("vend", case=False), "Tipo"] = "Entrada"
caixa["Tipo"].value_counts(dropna=False)   # deve mostrar só Entrada e Saída 
```

```python
#verificação de duplicadas
#aqui linhas iguais podem ser uma cópia por engano, ou dois lançamentos reais 
# a analise tem que ser mais cuidadosa

colunas_chave = ["Data", "Descrição", "Tipo", "Valor"]

dup = caixa[caixa.duplicated(subset=colunas_chave, keep=False)]
print("Linhas envolvidas em duplicidade:", len(dup))
print("Por aba:")
print(dup["aba_origem"].value_counts())

display(dup.sort_values(["Data", "ordem_original"]))
```

```python
# pq todas as duplicadas são de uma unica semana? 
# a distancia entre as pilhas é igual ?
dup = dup.copy()
dup["grupo"] = dup.groupby(colunas_chave).ngroup()
dup["qtd_no_grupo"] = dup.groupby("grupo")["grupo"].transform("size")

display(dup.sort_values(["grupo", "ordem_original"])
           [["grupo", "qtd_no_grupo", "ordem_original", "aba_origem"] + colunas_chave + ["Quem"]])
```

```python
# quantas linhas de vendas existem em um dia normal ? 

vendas = caixa[caixa["Descrição"].str.contains("vend", case=False, na=False)]
print(vendas.groupby(vendas["Data"].dt.date).size().value_counts())

#o normal é uma linha de venda por dia, possivelmente é anotado o total do dia em uma unica linha
#apenas 6 dias tem 2 linhas de vendas
```

```python
por_dia = vendas.groupby(vendas["Data"].dt.date).size()
print(por_dia[por_dia > 1])
# os dias com duas vendas correspondem aos dias duplicados
```

```python
#exclusão dessas duplicadas
antes = len(caixa)

copias = dup.sort_values("ordem_original").duplicated(subset=colunas_chave, keep="first")
idx_copias = dup.sort_values("ordem_original")[copias].index

caixa = caixa.drop(index=idx_copias)

print("Linhas removidas:", antes - len(caixa))
print("Duplicadas restantes:", caixa.duplicated(subset=colunas_chave).sum())
```

```python
#verificação de lacunas de datas

# todos os dias do período × dias com algum registro no caixa
todos = pd.date_range(caixa["Data"].min(), caixa["Data"].max(), freq="D")
com_registro = caixa["Data"].dt.normalize().unique()
sem_registro = todos.difference(com_registro)

print("Dias sem registro:", len(sem_registro))
print(pd.Series(sem_registro.day_name()).value_counts())
```

```python
dias_pt = {
    "Monday": "segunda", "Tuesday": "terça", "Wednesday": "quarta",
    "Thursday": "quinta", "Friday": "sexta", "Saturday": "sábado", "Sunday": "domingo",
}

sem = pd.DataFrame({"Data": sem_registro})
sem["dia_semana"] = sem["Data"].dt.day_name().map(dias_pt)
sem["mes"] = sem["Data"].dt.to_period("M")

# quantos dias sem registro por mês e por dia da semana
display(pd.crosstab(sem["mes"], sem["dia_semana"]))

# a lista completa, data a data
pd.set_option("display.max_rows", 200)
display(sem)
```
