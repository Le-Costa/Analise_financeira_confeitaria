```python
import pandas as pd
import numpy as np
path = r'C:\Users\Aluno\Documents\Analise_financeira_confeitaria\Dados\tratados'
caixa =pd.read_csv(path+r'\caixa_tratado.csv', sep=";", decimal=",", encoding="utf-8-sig", parse_dates=["Data"], dayfirst=True)
```

```python
ext = pd.read_csv(path+r'\extratos_tratados.csv', sep=";", decimal=",",
                  encoding="utf-8-sig", parse_dates=["Data"], dayfirst=True)
ext
```

```python
ext["Descrição"].unique()
```

```python
#separação por familia
partes = ext["Descrição"].str.split(" - ", n=1, expand=True)
ext["operacao"] = partes[0].str.strip()
ext["contraparte"] = partes[1].str.strip()

# tira números e códigos do fim da operação (parcela 01/24, TX885500...)
ext["operacao"] = ext["operacao"].str.replace(r"\s+(\d+/\d+|TX\d+)$", "", regex=True)

resumo = (ext.groupby(["operacao", "contraparte"], dropna=False)
             .agg(qtd=("Valor", "size"), total=("Valor", "sum"))
             .sort_values("operacao"))
print("Famílias de operação:", ext["operacao"].nunique())
print(ext["operacao"].value_counts())
display(resumo)
```

```python
enviados = ext[ext["operacao"] == "PIX ENVIADO"].copy()
enviados["dia"] = enviados["Data"].dt.day

perfil = (enviados.groupby("contraparte")
            .agg(qtd=("Valor", "size"),
                 total=("Valor", "sum"),
                 valores_diferentes=("Valor", "nunique"),
                 valores_mais_comuns=("Valor", lambda v: v.value_counts().head(3).index.tolist()),
                 dias_mais_comuns=("dia", lambda d: d.value_counts().head(3).index.tolist()),
                 primeiro=("Data", "min"),
                 ultimo=("Data", "max"))
            .sort_values("qtd", ascending=False))
display(perfil)
```

```python
mari = enviados[enviados["contraparte"] == "MARIANA SOUZA OLIVEIRA"]

# quantas vezes cada valor aparece
print(mari["Valor"].value_counts())

# o valor de 3.500 aparece uma vez por mês? em que dia?
p3500 = mari[mari["Valor"] == -3500]
print(p3500.groupby(p3500["Data"].dt.to_period("M")).size().value_counts())
print(p3500["dia"].value_counts().head())
```

```python
for nome in ["ANA PAULA RIBEIRO", "LETICIA ALVES", "BEATRIZ SANTOS"]:
    p = enviados[enviados["contraparte"] == nome]
    print(nome)
    print(p.groupby(p["Valor"])["Data"].apply(lambda d: sorted(d.dt.strftime("%m/%Y").unique())).to_string(), "\n")

#Essas mulheres são funcionarias. Os pagamentos indicaam pagamento de adiaantamento e salário, e décimo terceiro no fim do ano 
```

```python
#aonde esta o pafamento da JU? 
# algum lançamento no banco que mencione Juliana/Ju?
print(ext[ext["Descrição"].str.contains("JULIANA|JU ", case=False, na=False)]["Descrição"].unique())

# alguma saída no caixa físico para ela?
print(caixa[caixa["Descrição"].str.contains("ju", case=False, na=False)]["Descrição"].unique())
```

```python
ja_decididos = ["PIX ENVIADO"]   # família que você já classificou

falta = ext[~ext["operacao"].isin(ja_decididos)].copy()
falta["sentido"] = np.where(falta["Valor"] > 0, "Entrada", "Saída")

resumo_falta = (falta.groupby(["sentido", "operacao"])
                  .agg(qtd=("Valor", "size"),
                       total=("Valor", "sum"),
                       contrapartes=("contraparte", "nunique"))
                  .sort_values(["sentido", "qtd"], ascending=[True, False]))
display(resumo_falta)
```
