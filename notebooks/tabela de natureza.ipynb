```python
#a ideia é que a linha do extrato seja a somaa daas vends em pix da maaquininha daquele dia. As duas bases devem mostrar o mesmo valor

# 1. maquininha: vendas em pix (sem as canceladas), somadas por data de recebimento
maq_pix = maq[(maq["Tipo pagamento"] == "Pix") & (maq["Status"] != "Cancelada")]   # ajuste os rótulos ao seu de-para
maq_dia = (maq_pix.groupby(maq_pix["Data recebimento"].dt.normalize())
                  .agg(liquido_maq=("Valor liquido", "sum"), bruto_maq=("Valor bruto", "sum"))
                  .round(2))

# 2. extrato: repasses de pix da maquininha, somados por dia
ext_pix = ext[ext["Descrição"] == "VENDAS PIX MAQUININHA - REPASSE"]
ext_dia = ext_pix.groupby(ext_pix["Data"].dt.normalize())["Valor"].sum().round(2).rename("repasse_banco")

# 3. lado a lado
comp = pd.concat([maq_dia, ext_dia], axis=1)
comp["dif"] = (comp["repasse_banco"] - comp["liquido_maq"]).round(2)

print("Dias comparados:", len(comp))
print("Dias com diferença:", (comp["dif"].abs() > 0.01).sum())
print("Dias só na maquininha:", comp["repasse_banco"].isna().sum())
print("Dias só no banco:", comp["liquido_maq"].isna().sum())
display(comp[comp["dif"].abs() > 0.01].head(10))
```

```python
#mesma conferência para débito e crédito
def conferir_repasse(tipo_pag, descricao_extrato, fim_extrato="2026-10-31"):
    m = maq[(maq["Tipo pagamento"] == tipo_pag) & (maq["Status"] != "Cancelada")]
    print(f"--- {tipo_pag} --- vendas encontradas na maquininha: {len(m)}")
    m_dia = m.groupby(m["Data recebimento"].dt.normalize())["Valor liquido"].sum().round(2).rename("liquido_maq")

    e = ext[ext["Descrição"] == descricao_extrato]
    e_dia = e.groupby(e["Data"].dt.normalize())["Valor"].sum().round(2).rename("repasse_banco")

    comp = pd.concat([m_dia, e_dia], axis=1)
    ambos = comp.dropna()
    so_maq = comp[comp["repasse_banco"].isna()]
    so_banco = comp[comp["liquido_maq"].isna()]

    print("Dias com os dois lados:     ", len(ambos))
    print("  desses, com diferença:    ", ((ambos["repasse_banco"] - ambos["liquido_maq"]).abs() > 0.01).sum())
    print("Dias só na maquininha:      ", len(so_maq),
          f"(desses, {(so_maq.index > pd.Timestamp(fim_extrato)).sum()} depois de {fim_extrato})")
    print("Dias só no banco:           ", len(so_banco))
    print("A receber depois do fim:    ", so_maq.loc[so_maq.index > pd.Timestamp(fim_extrato), "liquido_maq"].sum().round(2), "\n")
    return comp

comp_deb = conferir_repasse("Debito", "VENDAS CARTAO DEBITO - REPASSE")
comp_cred = conferir_repasse("Credito", "VENDAS CARTAO CREDITO - REPASSE")

#grande parte das venda em credito nao foram encontradas no extrato bancario, e nenhuma no debito
```

```python
caixa = pd.read_csv(path + r"\caixa_tratado.csv", sep=";", decimal=",",
                    encoding="utf-8-sig", parse_dates=["Data"], dayfirst=True)   # ajuste o path

# 1. depósitos no caixa (saídas da gaveta)
dep_caixa = caixa[(caixa["Tipo"] == "Saída")
                  & caixa["Descrição"].str.contains("dep", case=False, na=False)].copy()
print("Descrições encontradas no caixa:", dep_caixa["Descrição"].unique())
dep_caixa["valor_abs"] = dep_caixa["Valor"].abs().round(2)

# 2. depósitos no extrato (entradas no banco)
dep_banco = ext[ext["Descrição"] == "DEPOSITO EM DINHEIRO"].copy()
dep_banco["valor_abs"] = dep_banco["Valor"].round(2)

print("Depósitos no caixa:", len(dep_caixa), "| total:", dep_caixa["valor_abs"].sum())
print("Depósitos no banco:", len(dep_banco), "| total:", dep_banco["valor_abs"].sum())

# 3. cruzamento por data + valor
cruz = dep_banco[["Data", "valor_abs"]].merge(dep_caixa[["Data", "valor_abs", "Descrição"]],
                                              on=["Data", "valor_abs"], how="outer", indicator=True)
print(cruz["_merge"].value_counts())
display(cruz[cruz["_merge"] != "both"].sort_values("Data"))
```

```python
#categorização da planilha de caixa

caixa["desc_limpa"] = caixa["Descrição"].str.strip().str.lower()

resumo_caixa = (caixa.groupby(["Tipo", "desc_limpa"], dropna=False)
                  .agg(qtd=("Valor", "size"),
                       total=("Valor", "sum"),
                       categorias_2026=("Categoria", lambda c: sorted(c.dropna().unique().tolist())),
                       abas=("aba_origem", lambda a: sorted(a.unique().tolist())))
                  .sort_values(["Tipo", "qtd"], ascending=[True, False]))
pd.set_option("display.max_rows", 200)
display(resumo_caixa)
```

```python
regras_caixa = {
    r"vend":               "Vendas em dinheiro",
    r"ovo|feira|fruta":    "Matéria-prima",
    r"moto ?boy|entrega":  "Entregas",
    r"\bdep":              "Transferência (caixa → banco)",
    r"g[aá]s|botij":       "Contas de consumo",
    r"limpeza":             "Despesas gerais",
    r"mari":               "Retirada da sócia",
}

caixa["categoria_final"] = None
for padrao, cat in regras_caixa.items():
    caixa.loc[caixa["categoria_final"].isna()
              & caixa["Descrição"].str.contains(padrao, case=False, na=False, regex=True),
              "categoria_final"] = cat

print(caixa["categoria_final"].value_counts(dropna=False))
display(caixa[caixa["categoria_final"].isna()]["Descrição"].value_counts())
```
